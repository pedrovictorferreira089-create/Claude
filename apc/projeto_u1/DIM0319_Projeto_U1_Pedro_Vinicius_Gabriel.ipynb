### 4.1 Estatística de colunas numéricas

**Pergunta 1 — Qual é a produtividade média da fábrica e como ela se compara com a meta?**

### 4.3 Criar uma nova coluna
Criem pelo menos uma coluna a partir de outras (ex.: `df['nova'] = df['a'] + df['b']`).

Vamos criar 3 colunas novas:
- `setor`: o nome do setor corrigido, em português (feito com `for` + `if`);
- `produtividade_pct`: a produtividade real em porcentagem;
- `bateu_meta`: `True` se a equipe alcançou a meta naquele dia.

```python
# Coluna 'setor': percorremos cada nome do setor e decidimos o nome correto
setores = []
for nome in df['department']:
    if nome == 'sweing':
        setores.append('costura')
    else:
        setores.append('acabamento')   # 'finishing' e 'finishing ' viram o mesmo setor

df['setor'] = setores
print(df['setor'].value_counts())
```

### 4.4 Usando NumPy
Transformem uma coluna em array e calculem algo (ex.: `np.mean`, `np.std`), **ou** usem uma máscara booleana em um array.

**Pergunta 3 — Com que frequência as equipes batem a meta, e quanto a produtividade varia?**

```python
produtividade = np.array(df['actual_productivity'])
meta = np.array(df['targeted_productivity'])

print('Media:', round(np.mean(produtividade), 3))
print('Mediana:', round(np.median(produtividade), 3))
print('Desvio padrao:', round(np.std(produtividade), 3))
```

```python
# Mascaras booleanas: True/False para cada registro
bateu = produtividade >= meta
print('Registros que bateram a meta:', np.sum(bateu), 'de', len(produtividade))
print('Percentual que bateu a meta:', round(np.mean(bateu) * 100, 1), '%')
print()

acima_100 = produtividade > 1
print('Registros acima de 100%:', np.sum(acima_100))

criticos = produtividade < 0.5
print('Registros criticos (abaixo de 50%):', np.sum(criticos))
print('Media so dos criticos:', round(np.mean(produtividade[criticos]), 3))
```

Resposta: cerca de 73% dos registros bateram a meta (875 de 1197). O desvio padrão de ≈ 0,17
(17 pontos percentuais) é alto para uma média de 0,73. Há 37 registros acima de 100%, e 133 registros críticos
(abaixo de 50%), são eles que puxam a média para baixo.

### 4.5 Uma função da equipe
Escrevam **uma função** (`def`) que organize um cálculo que vocês repetem (ex.: uma que receba uma coluna e devolva um resumo).

Criamos duas funções:
- `resumo(valores)`: recebe uma coluna e devolve um dicionário com média, desvio padrão, mínimo e máximo;
- `taxa_meta(tabela)`: recebe uma tabela (ou um pedaço dela) e devolve o % de registros que bateram a meta.

```python
def resumo(valores):
    """Recebe uma coluna de numeros e devolve um dicionario com o resumo."""
    valores = np.array(valores)
    return {
        'media': round(float(np.mean(valores)), 3),
        'desvio': round(float(np.std(valores)), 3),
        'minimo': round(float(np.min(valores)), 3),
        'maximo': round(float(np.max(valores)), 3),
    }


def taxa_meta(tabela):
    """Devolve o percentual de registros da tabela que bateram a meta."""
    return round(float(np.mean(tabela['bateu_meta'])) * 100, 1)


# Testando com a fabrica inteira
print('Fabrica toda:', resumo(df['actual_productivity']))
print('Bateram a meta:', taxa_meta(df), '%')
```

**Pergunta 4 — Qual setor é mais produtivo e mais estável: costura ou acabamento?**

```python
costura = df[df['setor'] == 'costura']
acabamento = df[df['setor'] == 'acabamento']

print('COSTURA')
print('  Produtividade:', resumo(costura['actual_productivity']))
print('  Bateram a meta:', taxa_meta(costura), '%')
print('  Operarios por equipe (media):', round(costura['no_of_workers'].mean(), 1))
print()
print('ACABAMENTO')
print('  Produtividade:', resumo(acabamento['actual_productivity']))
print('  Bateram a meta:', taxa_meta(acabamento), '%')
print('  Operarios por equipe (media):', round(acabamento['no_of_workers'].mean(), 1))
```

**Resposta:** a produtividade média é parecida (≈ 72% na costura e ≈ 75% no acabamento), mas o
**acabamento é muito mais instável**: desvio padrão maior e só **≈ 60% dos registros batem a meta**, contra **≈ 83%
na costura**. A costura trabalha com equipes grandes (≈ 52 operários) e o acabamento com equipes pequenas (≈ 10).

### 4.6 Laço e condicional
Em alguma etapa, usem um `for` percorrendo dados e um `if` para decidir algo (ex.: contar quantos registros passam de um limite).

**Pergunta 5 — Quais equipes ficam acima e quais ficam abaixo da meta?**

Para cada uma das 12 equipes: calculamos a produtividade média e a meta média, decidimos se ela ficou acima ou
abaixo da meta e guardamos a melhor e a pior equipe.

```python
equipes = []
medias = []
metas = []
situacoes = []

acima = 0
melhor_equipe = 0
melhor_media = 0
pior_equipe = 0
pior_media = 1

for equipe in range(1, 13):
    dados_equipe = df[df['team'] == equipe]
    media = round(dados_equipe['actual_productivity'].mean(), 3)
    meta_media = round(dados_equipe['targeted_productivity'].mean(), 3)

    if media >= meta_media:
        situacao = 'acima da meta'
        acima = acima + 1
    else:
        situacao = 'abaixo da meta'

    if media > melhor_media:
        melhor_media = media
        melhor_equipe = equipe
    if media < pior_media:
        pior_media = media
        pior_equipe = equipe

    equipes.append(equipe)
    medias.append(media)
    metas.append(meta_media)
    situacoes.append(situacao)

ranking = pd.DataFrame({
    'equipe': equipes,
    'produtividade_media': medias,
    'meta_media': metas,
    'situacao': situacoes
})
print(ranking)
print()
print('Equipes acima da meta:', acima, 'de 12')
print('Melhor equipe:', melhor_equipe, '| media', melhor_media)
print('Pior equipe:', pior_equipe, '| media', pior_media)
```

**Resposta:** **metade das equipes (6 de 12) fica, na média, abaixo da meta** — as equipes 6, 7, 8, 9, 10 e 11.
As equipes 1 a 5 e a 12 ficam acima. A **melhor é a equipe 1** (≈ 82%) e a **pior é a equipe 7** (≈ 67%). Como todas
as equipes têm metas parecidas (entre 0,67 e 0,77), a diferença está no desempenho, e não em metas mais difíceis.

**Pergunta 6 — O incentivo financeiro está ligado a uma produtividade maior?**

Primeiro vemos em quantos registros de cada setor houve incentivo:

```python
print('Costura com incentivo:', np.sum(costura['incentive'] > 0), 'de', len(costura))
print('Acabamento com incentivo:', np.sum(acabamento['incentive'] > 0), 'de', len(acabamento))
```

Quase só a costura recebe incentivo. Então comparamos **dentro da costura** os registros com e sem incentivo
(reaproveitando as funções `resumo` e `taxa_meta`):

```python
com_incentivo = costura[costura['incentive'] > 0]
sem_incentivo = costura[costura['incentive'] == 0]

print('Costura COM incentivo (', len(com_incentivo), 'registros )')
print('  ', resumo(com_incentivo['actual_productivity']))
print('   Bateram a meta:', taxa_meta(com_incentivo), '%')
print()
print('Costura SEM incentivo (', len(sem_incentivo), 'registros )')
print('  ', resumo(sem_incentivo['actual_productivity']))
print('   Bateram a meta:', taxa_meta(sem_incentivo), '%')
```

**Resposta:** sim. Na costura, quando houve incentivo a produtividade média foi de **≈ 76%**; sem incentivo,
caiu para **≈ 53%**. A taxa de metas batidas despenca de **≈ 92%** para **≈ 36%**. É a maior diferença que encontramos na base.
(Cuidado: isso mostra uma **relação**, não prova que o incentivo *causa* a alta — pode ser que o incentivo só seja pago
quando a equipe vai bem.)
